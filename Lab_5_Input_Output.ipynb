## Question 4: Read numbers and write formatted output

Use the program below to read from `numbers.txt` and write formatted results to `results.txt`.

When prompted, type:

```text
numbers.txt
results.txt
```

Note: the book shows you the original python formatting syntax, but you can use f-strings instead if you prefer. Both are fine, it is up to you which one you use.
Important! If you spell the file names wrong when you are typing them in to the prompt, you will get an error. The program will not know what file to open if you spell it incorrectly! If you run the program and it creates a new file with a different name, that means you misspelled the output file name. You can delete that file and run the program again.

```python

````python
# Question 4

# Prompt the user for the name of the input and output files.
inputFileName = input("Input file name: ")
outputFileName = input("Output file name: ")

# Open the input and output files.
infile = open(inputFileName, "r")
outfile = open(outputFileName, "w")

# Read the input and write the output.
total = 0.0
count = 0

line = infile.readline()
while line != "":
    value = float(line)
    outfile.write("%15.2f\n" % value)
    total = total + value
    count = count + 1
    line = infile.readline()

# Output the total and average.
outfile.write("%15s\n" % "--------")
outfile.write("Total: %8.2f\n" % total)

avg = total / count
outfile.write("Average: %6.2f\n" % avg)

# Close the files.
infile.close()
outfile.close()

print("Done. Open results.txt to check your work.")
````
